Kiểm tra tập dữ liệu huấn luyện.

```python
train_data = pd.read_csv("train.csv" ,  low_memory=False)
print(train_data.head())
```

```python
print(train_data.shape)
```

```python
print(train_data.info())
```

Concatenate the two datasets to form a complete dataset

```python
my_data = pd.merge( info_data , train_data , on = 'Store' , how = 'left')
my_data = my_data.sort_values(by = ['Store','Date'] , ascending = True)
print(my_data.head())
```

```python
print(my_data.shape)
```

# **Tiền xử lý dữ liệu** 

Xem thông tin dữ liệu.

```python
print(my_data.info())
```

```python
print(my_data.describe())
```

Kiểm tra dữ liệu thiếu.

```python
missing_values = my_data.isnull().sum()
print(missing_values)
```

Vì không có đặc trưng nào có lượng dữ liệu bị thiếu quá nhiều, tiến hành điền các giá trị còn thiếu cho tất cả các cột.

```python
my_data['CompetitionDistance'] = my_data['CompetitionDistance'].fillna(999999)
data_fill_zero = ['CompetitionOpenSinceMonth' , 'CompetitionOpenSinceYear' , 'Promo2SinceWeek' , 'Promo2SinceYear']
my_data[data_fill_zero] = my_data[data_fill_zero].fillna(0)
my_data['PromoInterval'] = my_data['PromoInterval'].fillna('missing')
new_missing_values = my_data.isnull().sum()
print(new_missing_values)
```

Kiểm tra trùng lặp dữ liệu.

```python
print("Số dữ liệu trùng lặp: " , my_data.duplicated().sum())
display( my_data[my_data.duplicated(keep=False)])
```

Kiểm tra dữ liệu cùng cửa hàng, cùng ngày nhưng thông tin khác nhau.

```python
dulicated_key = my_data.duplicated(subset=['Store' , 'Date'] , keep=False)
display(my_data.loc[dulicated_key].sort_values(['Store' , 'Date']) )
```

tiến hành loại bỏ các giá trị có khả năng gây nhiễu. Chúng ta xóa cột 'mở' vì nếu một cửa hàng đóng cửa, doanh thu của cửa hàng đó bằng 0; vì ngày đóng cửa bị loại trừ nên cột 'mở' sẽ chỉ chứa giá trị 1, khiến việc đào tạo mô hình trở nên vô dụng.

```python
my_data = my_data.drop('Open' , axis = 1)
my_data = my_data[my_data['Sales'] > 0 ]
print('Complete!')
```
