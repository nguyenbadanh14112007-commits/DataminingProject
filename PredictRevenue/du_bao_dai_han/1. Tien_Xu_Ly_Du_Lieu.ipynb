# **1. Tiền Xử Lý Dữ Liệu**

Đọc hai file `store.csv` và `train.csv`, ghép lại, xử lý giá trị thiếu, loại ngày đóng cửa và lưu dữ liệu sạch cho các notebook sau.

```python
from pathlib import Path

import pandas as pd

PROCESSED_DIR = Path('data/processed')
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
NO_COMPETITOR_DISTANCE = 999999
```

Khai báo các thư viện liên quan.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

kiểm tra dữ liệu về thông tin chuỗi cửa hàng.

```python
info_data = pd.read_csv("store.csv")
info_data.head()
```

```python
print(info_data.shape)
```

```python
print(info_data.info())
```

Kiểm tra dữ liệu của tập train.csv.

```python
train_data = pd.read_csv("train.csv" ,  low_memory=False)
train_data.head()
```

```python
print(train_data.info())
```

```python
my_data = pd.merge( info_data , train_data , on = 'Store' , how = 'left')
my_data = my_data.sort_values(by = ['Store','Date'] , ascending = True)
my_data.head()
```

```python
print(my_data.shape)
```

```python
print(my_data.info())
```

Kiểm tra dữ liệu bị thiếu.

```python
missing_values = my_data.isnull().sum()
print(missing_values)
```

Dữ liệu tương đối sạch, chỉ thiếu ở một vài đặc trưng nhất định nhưng không nhiều. Phần điền khuyết phụ thuộc vào mô hình sử dụng.

Kiểm tra dữ liệu trùng lặp.

```python
print('Số dòng trùng lặp hoàn toàn:', my_data.duplicated().sum())
print('Số cặp (Store, Date) bị lặp:', my_data.duplicated(subset=['Store', 'Date']).sum())
```

Chọn cột mốc thời gian phân chia dữ liệu nhằm phục vụ cho việc khám phá, khác thác dữ liệu.

```python
TRAIN_END = pd.Timestamp('2015-04-24')
VALID_END = pd.Timestamp('2015-06-12')
TRAIN_TEST_END = pd.Timestamp('2013-12-31')
VALID_TEST_END = pd.Timestamp('2014-02-19')
```

Lưu dữ liệu để tái sử dụng.

```python
my_data.to_csv(PROCESSED_DIR / 'rossmann_clean.csv', index=False)
print('Đã lưu', PROCESSED_DIR / 'rossmann_clean.csv')
```

```python
import json

config = {
    'TRAIN_END': TRAIN_END,
    'VALID_END': VALID_END,
    'TRAIN_TEST_END': TRAIN_TEST_END,
    'VALID_TEST_END': VALID_TEST_END,
}

with open(PROCESSED_DIR / 'config.json', 'w', encoding='utf-8') as f:
    json.dump({k: v.strftime('%Y-%m-%d') for k, v in config.items()}, f, indent=2)
print('Đã lưu', PROCESSED_DIR / 'config.json')
```

Đọc và lưu file test.

```python
test_data = pd.read_csv('test.csv')
test_data.head()
```

```python
my_test = pd.merge( test_data , info_data , on = 'Store' , how = 'left')
my_test.head()
```

Kiểm tra các cặp bị khuyết và điền khuyết.

```python
missing_values_test = my_test.isnull().sum()
print(missing_values_test)
```

Kiểm tra trùng lặp.

```python
print('Số dòng trùng lặp hoàn toàn:', my_test.duplicated().sum())
print('Số cặp (Store, Date) bị lặp:', my_test.duplicated(subset=['Store', 'Date']).sum())
```

lưu dữ liệu.
